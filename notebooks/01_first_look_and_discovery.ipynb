## 2. Discover a real truth yourself

The safest, most satisfying early wins are **things you find in the data** — un-leakable, and they *are* the core lesson. Run the three cells below. Each is ~10 lines of pandas and each overturns a common SEO belief.

Every number is **computed live from the shipped CSV** — nothing is hardcoded.

```python
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)
```

### Discovery A — "High search volume means more traffic." Does it?

```python
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Correlation between search_volume and impressions_90d: {corr:.3f}")
print("Near zero -> keyword search volume barely predicts the traffic a page actually gets.")
```

### Discovery B — the CTR cliff by position
Click-through rate is not flat: it collapses as you move down the results.

```python
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(ctr_by_pos.round(4).to_string())
ctr_by_pos.plot(kind="bar", title="Mean CTR by position tier (impressions >= 100)", ylabel="CTR");
```

### Discovery C — is longer content the lever?
Compare word count for **declining** vs **growing** pages.

```python
wc = df.groupby("trend_direction")["word_count"].median()
print(wc.round(0).to_string())
print("\n'down' vs 'up' pages have almost the same median word count -> length is not the lever.")
```

### Finding more info from discovery B (its fun so why *not*)

To find the `content_type` with the worst CTR at the same `position_tier`, we first group the `visible` DataFrame (which already filters for `impressions_90d >= 100`) by both `position_tier` and `content_type`. Then, we calculate the mean `ctr` for each of these groups.

After obtaining these grouped average CTRs, we'll `unstack()` the `content_type` to make it easier to compare within each `position_tier`. Finally, we find the `content_type` that corresponds to the minimum CTR for each `position_tier`.

```python
# Number of unique values in the 'content_type' column
unique_content_types = df['content_type'].nunique()
print(f"Number of unique content types: {unique_content_types}")

# Unique values using .unique()
print("\nUnique content types:")
print(df['content_type'].unique())

# Number of unique values in the 'position_tier' column
unique_content_types = df['position_tier'].nunique()
print(f"Number of unique position tiers: {unique_content_types}")

# Unique values using .unique()
print("\nUnique position tiers:")
print(df['position_tier'].unique())
```

```python
# Group by position_tier and content_type, then calculate the mean CTR and unstack content_type to make comparison across content_types easier for each position_tier
worst_ctr_by_pos_type = visible.groupby(['position_tier', 'content_type'])['ctr'].mean().unstack()

print("Average CTR by Position Tier and Content Type:")
display(worst_ctr_by_pos_type.round(4))


worst_content_type_per_position = worst_ctr_by_pos_type.idxmin(axis=1)

min_ctr_per_position = worst_ctr_by_pos_type.min(axis=1)

print("\nContent type with the worst CTR for each position tier:")
for pos_tier, content_type in worst_content_type_per_position.items():
    print(f"  In '{pos_tier}': '{content_type}' has the worst CTR (average CTR: {min_ctr_per_position[pos_tier]:.4f})")
```

Find another belief to test: does content_age_days relate to trend_direction? Does avg_position relate to ctr?

```python
age_and_trend_dir = df.groupby("trend_direction")["content_age_days"].mean().sort_values(ascending=False)
print(age_and_trend_dir.round(3).to_string())
age_and_trend_dir.plot(kind="bar", title="Avg age and trend direction.", ylabel="Age");
```

Content age varies across trend directions. Pages with an upward trend have the highest average content age, while pages with a downward trend have the lowest average content age, suggesting that content age may be associated with trend direction.
